## 3. Test NASA POWER API Connectivity

We run a lightweight query for a single location (Makandura) for the year 2020 to verify that the NASA POWER monthly weather API responds correctly.

```python
import requests

test_estate = "Makandura"
coords = config.ESTATES[test_estate]
params = {
    "latitude": coords["lat"],
    "longitude": coords["lon"],
    "start": "2020",
    "end": "2020",
    "parameters": ",".join(config.WEATHER_PARAMETERS),
    "community": "AG",
    "format": "JSON"
}

print(f"Testing NASA API connection for {test_estate}...")
try:
    response = requests.get(config.NASA_POWER_URL, params=params, timeout=15)
    if response.status_code == 200:
        print("[SUCCESS] NASA POWER API is reachable and returned data successfully!")
        data = response.json()
        print("Parameters returned:", list(data["properties"]["parameter"].keys()))
    else:
        print(f"[FAILED] NASA POWER API returned status code: {response.status_code}")
except Exception as e:
    print(f"[ERROR] Connectivity test failed: {str(e)}")
```
